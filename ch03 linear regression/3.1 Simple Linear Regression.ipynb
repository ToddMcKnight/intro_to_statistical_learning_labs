# 3.1 Simple Linear Regression

$$Y \approx \beta_0 + \beta_1 X$$
- Betas are the coefficients/parameters
- $\hat \beta_0$ and $\hat \beta_0$ are the model coefficients
- Can predict the new response $\hat y$
- In practice $\beta_0 \text{and} \beta_1$ are unknown so we use data to estimate them
- many ways to measure and minimize error but most common is least squares
- Let $\hat y_i = \hat \beta_0 + \hat \beta_1 x_i$ be the prediction for Y based on the $i$ value of X. $e_i = y_i - \hat y_i$ is the ith residual
- Residual squares sums $RSS = e_1^2 + e_2^2+...e_n^2$
- Using calculus we find the minimum RSS is found when:
$$\hat \beta_1 = \frac{\sum_{i=1}^n(x_i-\bar x)(y_i - \bar y)}{\sum_{i=1}^n(x_i-\bar x)^2}$$
$$\hat \beta_0 = \hat y - \hat \beta_1 \bar x$$

- Where $\hat y$ and $\hat x$ are the arithmetic mean of X and Y



## 3.1.2 Assessing the Accuracy of the Coefficient Estimates
- Recall we assume the true relationship is always $Y=f(X)+\epsilon$
- In this case we assume f is linear so we assume $Y=\beta_0 + \beta_1 X + \epsilon$
- We typically assume that error term is independent of X
- Recall from a statistics class that for an estimated to be unbiased it's expected value equals the true population value
- Just like for estimating mean we can calculate the standard error we can do the same for $\beta_0$ and $\beta_1$
$$\operatorname{SE}(\hat \beta_0)^2 = \sigma^2[\frac{1}{n}+\frac{\bar x^2}{\sum_{i=1}^n(x_i-\bar x)^2}]$$
$$\operatorname{SE}(\hat \beta_1)^2 = \frac{\sigma^2}{\sum_{i=1}^n(x_i-\bar x)^2}$$
- Where $\sigma^2 = \operatorname{Var}(\epsilon)$
- These formulas assume taht resideuals have common variance and are uncorrelated which is typically not true
- $\beta_1$ has lower standard error when there is more spread in $X$
- $\sigma$ is usually not known but is estimated as the residual standard error $RSE = \sqrt{RSS/(n-2)}$
- For both betas the 95% condidence interval is the estimate +/- 2 times the standard error
- From a hypothesis testing perspecting, no relationship between X and Y correscpods to $\beta_1 = 0$
- In practice compute t statistc $t = \frac{\hat \beta_1 - 0}{\operatorname{SE}(\hat \beta_1)}$
- Normal p-value idea p-value is the probability of seeing the results we'd see or more extreme given the null hypothesis is true

## 3.1.3 Assesing the Accuracy of the Model
- Tested using two statistics, the *residual standard error* and $R^2$ statistics
### Residual Standard Error
$$RSE = \sqrt{\frac{1}{n-2}RSS} = \sqrt{\frac{1}{n-2}\sum_{i=1}^n(y_i-\hat y_i)^2}$$
- RSE is the estimate of the standard deviation of $\epsilon$
- Considered a measure of *lack of fit*
### $R^2$ Statistic
$$R^2 = \frac{TSS-RSS}{TSS}$$
- Where TSS is the total sum of squares $TSS = \sum(y_i-\bar y)^2$
- TSS is the total variance in the response Y while RSS is the variance that is left after performing the regression
- $R^2$ is like the proportion of variation in $Y$ that is explained by $X$
- One advantage $R^2$ has over RSE is that $R^2 \in [0,1]$
- Recall the correlation is $$\operatorname{Cor}(X, Y) = \frac{\sum_{i=1}^{n} (x_i - \overline{x})(y_i - \overline{y})}{\sqrt{\sum_{i=1}^{n} (x_i - \overline{x})^2} \sqrt{\sum_{i=1}^{n} (y_i - \overline{y})^2}},$$
- Turns out that if we set correlation to little $r$ then $r^2$ = $R^2$ in the simple setting.
