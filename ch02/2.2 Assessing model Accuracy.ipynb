# Assessing Model Accuracy

## 2.2.1 Measuring the Quality of Fit
- Need to measure how well predictions match observed data
- Commonly use Mean Squared Error MSE 
$$MSE = \frac{1}{n} \sum_{i=1}^n (y_i-\hat f(x_i))^2$$
- $y_i is the real response and $\hat f$ is the predicted value
- MSE is computed using training data is called the training MSE
- In general we care about the the error on data not included in the training data
- Test MSE: $Ave(y_0-\hat f(x_0))^2$ where $(x_0,y_0)$ are new data points
- Trying to minimize test MSE only helps to a point and eventually has diminishing return: this is called overfitting


## 2.2.2 The Bias Variance Trade-Off
$$\operatorname{E}(y_0 - \hat f(x_0))^2 = \operatorname{Var}(\hat f(x_0)) + [\operatorname{Bias}(\hat f(x_0))]^2 + \operatorname{Var}(\epsilon)$$

- RHS is the expted test MSE for a new data point $x_0$
- Variance of $\hat f$ is the how much $\hat f$ varies with a different set of training data
- Bias of $\hat f$ is how much is lost in the process of approximating the real life process
- Generally, inflexible models have low variance and high bias and flexible models have high variance and low bias
- Bias tends to decrease faster than variance increases initially but as flexibility increases eventually the bias barely changes but variance rises a lot

## 2.2.3 The classification setting
Error training rate of classification problems
$$\frac{1}{n}\sum_{i=1}^n I(y_i \not = \hat y_i)$$
- I is the indicator function, 1 if true 0 is not
- Proportion of predictions that are correct by the model
- Test error rate is the average for new data 
### Bayes Classifier
 - The best classifier is the one that assigns each observation to the class with the higher probability
 - $\operatorname{Pr}(Y=j|X=x_0)$
 - The line where probabilities tie is called the Bayes decision boundary
### K-Nearest Neighbors (KNN)
- Given a new data point select the K- nearest neighbors and estimate the probability of class $i$ as the the proportion of the KNNs that are in class $i$
- Generally low values of K mean high flexibility and high values of K mean low flexibility
- Incidentally high values of K result in the barrier line being close to linaer
