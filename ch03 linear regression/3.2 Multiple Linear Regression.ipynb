# 3.2 Multiple Linear Regression

- When we have multiple independent variables we may want to employ all of them in a model
$$Y= \beta_0 + \beta_1 X_1 + \beta_2 X_2 + \cdots + \beta_p X_p + \epsilon$$

## 3.2.1 Estimating the Regression Coefficients
- Same estimation paradigm still applies with hats and such
- RSS is the still same sum of difference between true response and predicted response
- There can be a case where the simple regression leads to a statistically significant beta but the multi regression doesn't
- Why does adding in new variables mean that one field no longer has an effect?
    - It could be that field A is correlated with Field B and then Field B drives the response. So if field B is not present than field A will appear to have a large effect but one field B is present in the regression then it can take credit for it's work
    

## 3.2.2 Some important Questions
### Is there a relationship between the response and predictors?
- In the simple case, $H_0$ was that $B_1=0$ but in the multi case $H_0$ is that ALL of the slope parameters are 0 and the alternative is that *at least one $\beta_j$ is non zero$
- This needs an F-test 
$$F = \frac{(TSS-RSS)/p}{RSS/(n-p-1)}$$
- Where n is the number of observations and p is the number of predictors
- If the true relationship is linear than the expected value of the denominator is $\sigma$^2
- If the null hypthesis is true then the numerator is expected to be $\sigma^2$ but if the alternative is true than the numerator will be higher so the overall F statiststic will be greater than 1
- When we have a very large number of predictors it is very likely that the p value for a few of them will be significant by chance, but F statistic takes care of this
### Deciding on Important Variable
- Once we have verified that the F value is statistically significant the question is which variables are critical
- In an ideal world we test every setup and use some statistic to determine which is best for for a lot of variables this is unweildly
    - Forward Selection: Start with no variable and then do a simple regression and add the variable that results in the lowest RSS then consider which second variable would result in the smallest RSS for the two variable model... Go until a stopping rule is satisfied
    - Backward Selection: Start with all variables and remove the largest p-value, then refit and remove the next largest p value. Repeat until all p-values are below some threshold
    - Mixed Selctio: Start doing forward selection but if during the refit any selection has a p value too high we kick it out
### Model Fit
- In multi regression $R^2 = \operatorname{Cor}(Y,\hat Y)^2$. Correlation between the response and the fitted linear model
- R^2 always increases as more variable are added, even if the association to the response is weak
- $RSE = \sqrt{\frac{1}{n-p-1}RSS}$
### Predictions
- Confidence interval account for reducible error and bias 
- Prediction interval accounts for irreducable error as well
