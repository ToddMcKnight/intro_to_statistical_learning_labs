# 3.3 Other Considerations

## 3.3.1 Qualitative Predictiors
- Sometimes predictors are quantitive, region, marital status etc
### Predictors with Only Two Levels
- Incorporate into regression via indicator variable
- Does not have to be ${0,1}$. Value of $beta_i$ variable assoc will change relative to
  choices of indicator
- Could instead choose ${-1,1}$ or anything else
- Doesn't change p-value but changes intepretation
### Qualitative Predictors with More than Two Levels
- Can add more dummy variables to say if they are in any category
- Need one fewer dummy variables than number of categories because nothing
  selected represents the baseline
- When you have a bunch of dummy variables the same f-test idea applies where
  coincidental lineups can skew results
- There are other ways to encode qualitative predictors but they are all
  equavalent to dummy variables up to some idea of equal
  


## 3.3.2 Extensions of the Linear Model
- Linear model implcitly assume predictors and response are additive and linear
- Assoc bw X_j and Y does not depend on the other predictors in linear model
### Removing the Additive Assumption
- Perhaps one predictor increases the effictiveness of the other predictor
- *Interaction Effect* when predictors have synergy
-$$Y = \beta_{0} + \beta_{1}X_i + \beta_{2}X_2  +\beta_{3}X_1X_2$$
- Think about how you can factor $X_1$ out of $\beta_{1}+\beta_{2}X_2$ so that
  mx+b type term defines how much $Y$ changes with one unit $X_1$
- throughput as a function of workers and production lines
    - New lines with no workers should increase throughput
    - Kinda calculus mindset where $\beta_{3}$ is the increase is the amount
      that one unit increase increases the response
- Heirarchical principle
    - In the scenerio where the intercation term is statistically signifant but
      the base terms are not then we must also include the main terms
- Imagine one of the terms, $X_2$ is categorical, with no iteraction term the
  regression amounts to finding two parallel lines in the $X_1, Y$ space.
    - interaction term allows for the slope to be different
    - Always think about factoring out one of the terms and interpreting it as a
    the slope
