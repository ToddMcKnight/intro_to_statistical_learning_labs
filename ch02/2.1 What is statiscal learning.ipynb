# What is statiscal learning


- Input variables
    - Typically denoted with $X$ and a subscript for which field eg $X_1, X_2, X_3$ are for field 1 field 2 field 3
    - AKA predictors, independent variablce, features, just variables
- Response
    - Typically denoted with $Y$
    - AKA dependent variables
- $$Y=f(X)+\epsilon$$
    - $Y$ is the response
    - $f(X)$ is some function of the predictors
    - $\epsilon$ is an error term with $\mu=0$ 
    - In principle if $\epsilon$ does not have mean zero than we can alter $f(X)$ by adding a constant term to move $\epsilon$ to have mean 0
- Primary goal of statiscal learning is to estimate $f$, denoated by $\hat f$
    


## 2.1.1 Why estimate $f$
Prediction and inference
- Prediction
    - $\hat f$ is treated as a black box. Exact form is irrelevent provided predictions are accurate
    - Reducable Error: Error that can be reduced by improving the accuracy of $\hat f$ as an estimate
    - Irreducable error: Error that cannot be reduced because it comes from $\epsilon$
- Inference
    - Goal is to understand the relationship between $X$ and $Y$
- Some scenerios fall under both
    - Model home values based home attributes
    - Want to predict what this homes values will be but also understand what should be done to improve home value
    


## 2.1.2 How do we estimate $f$?
Training data: Observations made that will be used to train the model
- $x_ij$ is used to represent the $j$'th predictor of the $i$'th observation
- $x_i$ is a vector of all fields for one observation
- Parametric models
    - First assume the functional form of $f$ then minimize the error over the paramters of the form
- Non-Paramteric Methods
    - Function that gets close to the data points as possible while maintaining an arbitrary level of 'smoothness'

## 2.1.3 Accuracy - Intepretability Tradeoff
- Sometime we choose a less flexible approach because it allows for more intuition
- Nueral networks have high accuracy but low intepretability


## 2.1.4 Supervised vs Unsupervised
- Supervised
    - Have a known output and we can measure how accurately the model predicts the output
- Unsupervised
    - No response variable
    - Cluster analysis group observations together
    

## 2.1.5 Regression vs Classification
- Regression is used when the response is quantitative
- Classification is used when the response is qualitative
